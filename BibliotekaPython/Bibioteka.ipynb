```python
from datetime import date, timedelta

class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True

    def __str__(self):
        status = "Available" if self.is_available else "Loaned out"
        return f"{self.title} by {self.author} (ISBN: {self.isbn}) - {status}"

class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.loans = []

    def __str__(self):
        return f"Member: {self.name} (ID: {self.member_id})"

class Loan:
    def __init__(self, book, member, loan_date=None, return_date=None):
        self.book = book
        self.member = member
        self.loan_date = loan_date or date.today()
        self.due_date = self.loan_date + timedelta(days=14)
        self.return_date = return_date

    def is_overdue(self):
        return self.return_date is None and date.today() > self.due_date

    def __str__(self):
        if self.return_date:
            status = f"Returned on: {self.return_date}"
        else:
            status = f"Due: {self.due_date}"
        return f"{self.book.title} loaned to {self.member.name} ({status})"

class Library:
    def __init__(self):
        self.books = []
        self.members = []
        self.loans = []

    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def loan_book(self, book_isbn, member_id):
        book = next((b for b in self.books if b.isbn == book_isbn and b.is_available), None)
        member = next((m for m in self.members if m.member_id == member_id), None)

        if book and member:
            loan = Loan(book, member)
            book.is_available = False
            member.loans.append(loan)
            self.loans.append(loan)
            print(f"Loan created: {loan}")
        else:
            print("Loan failed: book not available or member not found")

    def return_book(self, book_isbn):
        loan = next((l for l in self.loans if l.book.isbn == book_isbn and l.return_date is None), None)
        if loan:
            loan.return_date = date.today()
            loan.book.is_available = True
            print(f"Book returned: {loan.book.title}")
        else:
            print("Return failed: active loan not found")

    def list_books(self):
        print("\nLibrary books:")
        for book in self.books:
            print(book)

    def list_overdue_loans(self):
        print("\nOverdue loans:")
        found = False
        for loan in self.loans:
            if loan.is_overdue():
                print(loan)
                found = True
        if not found:
            print("No overdue loans found.")

    def list_loans(self):
        print("\nAll current and past loans:")
        for loan in self.loans:
            print(loan)


# === Główna część programu ===

library = Library()

# Dodanie książek
books = [
    Book("1984", "George Orwell", "12345"),
    Book("Dune", "Frank Herbert", "67890"),
    Book("To Kill a Mockingbird", "Harper Lee", "10001"),
    Book("Pride and Prejudice", "Jane Austen", "10002"),
    Book("The Great Gatsby", "F. Scott Fitzgerald", "10003"),
    Book("Moby-Dick", "Herman Melville", "10004"),
    Book("Brave New World", "Aldous Huxley", "10005"),
    Book("Crime and Punishment", "Fyodor Dostoevsky", "10006"),
    Book("War and Peace", "Leo Tolstoy", "10007"),
    Book("The Catcher in the Rye", "J.D. Salinger", "10008"),
    Book("Jane Eyre", "Charlotte Brontë", "10009"),
    Book("Wuthering Heights", "Emily Brontë", "10010"),
    Book("The Hobbit", "J.R.R. Tolkien", "10011"),
    Book("Fahrenheit 451", "Ray Bradbury", "10012"),
    Book("The Lord of the Rings", "J.R.R. Tolkien", "10013"),
    Book("The Picture of Dorian Gray", "Oscar Wilde", "10014"),
    Book("Dracula", "Bram Stoker", "10015"),
    Book("Frankenstein", "Mary Shelley", "10016"),
    Book("Anna Karenina", "Leo Tolstoy", "10017"),
    Book("Les Misérables", "Victor Hugo", "10018"),
    Book("Don Quixote", "Miguel de Cervantes", "10019"),
    Book("The Brothers Karamazov", "Fyodor Dostoevsky", "10020"),
    Book("The Count of Monte Cristo", "Alexandre Dumas", "10021"),
    Book("The Stranger", "Albert Camus", "10022"),
    Book("One Hundred Years of Solitude", "Gabriel García Márquez", "10023"),
    Book("The Metamorphosis", "Franz Kafka", "10024"),
    Book("Slaughterhouse-Five", "Kurt Vonnegut", "10025"),
    Book("Beloved", "Toni Morrison", "10026"),
    Book("Lolita", "Vladimir Nabokov", "10027"),
    Book("A Tale of Two Cities", "Charles Dickens", "10028"),
    Book("The Old Man and the Sea", "Ernest Hemingway", "10029"),
    Book("The Sun Also Rises", "Ernest Hemingway", "10030")
]

for book in books:
    library.add_book(book)

# Dodanie członków
members = [
    Member("Anna Kowalska", "M001"),
    Member("Agnieszka Nowak", "AN012"),
    Member("Andrzej Rybicki", "AR013"),
    Member("Wioletta Olszewska", "WO023"),
    Member("Tomasz Jasicki", "TJ025"),
    Member("Helena Boznańska", "HB045")
]

for member in members:
    library.add_member(member)

# Przykładowe wypożyczenia
library.loan_book("12345", "M001")     # 1984 -> Anna
library.loan_book("10001", "AN012")    # To Kill a Mockingbird -> Agnieszka
library.loan_book("10002", "AR013")    # Pride and Prejudice -> Andrzej
library.loan_book("10003", "WO023")    # The Great Gatsby -> Wioletta
library.loan_book("10004", "TJ025")    # Moby-Dick -> Tomasz
library.loan_book("10005", "HB045")    # Brave New World -> Helena

# Zwrócenie jednej książki
library.return_book("10003")           # Wioletta zwraca "The Great Gatsby"

# Wyświetlanie stanu biblioteki
library.list_books()
library.list_loans()
library.list_overdue_loans()
```
